```python
import soundfile as sf
import numpy as np
from datetime import datetime, timedelta

RAW_WAV = "/Users/qian/KWF/raw_recordings/20250319_180002.WAV"

# --- Read file metadata ---
info = sf.info(RAW_WAV)
sample_rate = info.samplerate
total_frames = info.frames
duration_sec = total_frames / sample_rate
print(f"File: {RAW_WAV}")
print(f"Sample rate: {sample_rate} Hz")
print(f"Duration: {duration_sec/3600:.2f} hours  ({total_frames:,} frames)")

# --- Stream through the file in 3-second windows, compute RMS per window ---
window_sec = 3.0
window_frames = int(window_sec * sample_rate)

print(f"\nComputing RMS per {window_sec}-second window...")
rms_values = []
with sf.SoundFile(RAW_WAV) as f:
    while True:
        block = f.read(window_frames, dtype="float32")
        if len(block) < window_frames:
            break
        if block.ndim > 1:
            block = block.mean(axis=1)
        rms = np.sqrt(np.mean(block ** 2))
        rms_values.append(rms)

rms_values = np.array(rms_values)
print(f"Total 3-second windows: {len(rms_values):,}")
print(f"\nRMS distribution:")
print(f"  min:     {rms_values.min():.6f}")
print(f"  5th %:   {np.percentile(rms_values, 5):.6f}")
print(f"  median:  {np.median(rms_values):.6f}")
print(f"  95th %:  {np.percentile(rms_values, 95):.6f}")
print(f"  max:     {rms_values.max():.6f}")

# --- Find continuous low-energy stretches ---
quiet_threshold = np.percentile(rms_values, 5)
is_quiet = rms_values <= quiet_threshold

min_run_windows = int(120 / window_sec)   # 2-minute minimum
runs = []
i = 0
while i < len(is_quiet):
    if is_quiet[i]:
        j = i
        while j < len(is_quiet) and is_quiet[j]:
            j += 1
        if (j - i) >= min_run_windows:
            runs.append((i, j, rms_values[i:j].mean()))
        i = j
    else:
        i += 1

print(f"\nQuiet threshold (bottom 5% of this recording): {quiet_threshold:.6f}")
print(f"Found {len(runs)} continuous quiet stretches of >= 2 minutes")

# --- Report the top 10 quietest stretches ---
import os
fname = os.path.basename(RAW_WAV)
date_str, time_str = fname.split(".")[0].split("_")
rec_start = datetime.strptime(date_str + time_str, "%Y%m%d%H%M%S")

runs_sorted = sorted(runs, key=lambda r: r[2])

print(f"\nTop 10 quietest sustained stretches:")
print(f"{'#':>3} {'Start time (clock)':>22} {'Offset (h:m:s)':>16} {'Duration':>11} {'Mean RMS':>12}")
for n, (start_idx, end_idx, mean_rms) in enumerate(runs_sorted[:10], start=1):
    offset_sec = start_idx * window_sec
    clock = rec_start + timedelta(seconds=offset_sec)
    dur_sec = (end_idx - start_idx) * window_sec
    h = int(offset_sec // 3600)
    m = int((offset_sec % 3600) // 60)
    s = int(offset_sec % 60)
    print(f"{n:>3} {clock.strftime('%Y-%m-%d %H:%M:%S'):>22} {h:>2}:{m:02d}:{s:02d}        {dur_sec/60:>5.1f} min  {mean_rms:>12.6f}")
```

```python
from IPython.display import display, Audio

def spot_check(offset_sec, duration_sec=30):
    """Seek into the raw WAV and play a segment — no full-file load."""
    start_frame = int(offset_sec * info.samplerate)
    read_frames = int(duration_sec * info.samplerate)

    with sf.SoundFile(RAW_WAV) as f:
        f.seek(start_frame)
        audio = f.read(read_frames, dtype="float32")

    if audio.ndim > 1:
        audio = audio.mean(axis=1)

    clock = rec_start + timedelta(seconds=offset_sec)
    print(f"Offset {offset_sec:,}s → {clock.strftime('%Y-%m-%d %H:%M:%S')}  ({duration_sec}s)")
    display(Audio(audio, rate=info.samplerate))

# Candidate #1 — quietest  (05:11:44, 6.6 min stretch)
spot_check(40302)

# Candidate #2             (05:09:02, 2.1 min stretch)
spot_check(40140)

# Candidate #3             (04:50:56, 2.6 min stretch)
spot_check(39054)

# Flatness candidate #3 — 04:07:44, 60.7 min
spot_check(36462)

# Flatness candidate #6 — 03:21:02, 46.6 min
spot_check(33660)
```

```python
# Spectral flatness scan — finds background candidates across all times of day,
# not just quiet pre-dawn windows. High flatness = broadband noise (wind, rain,
# ambient hiss). Low flatness = tonal structure (bird calls, frogs, insects).

window_sec = 3.0
window_frames = int(window_sec * info.samplerate)

print(f"Computing spectral flatness per {window_sec}-second window...")
flatness_values = []

with sf.SoundFile(RAW_WAV) as f:
    while True:
        block = f.read(window_frames, dtype="float32")
        if len(block) < window_frames:
            break
        if block.ndim > 1:
            block = block.mean(axis=1)

        # Wiener entropy: geometric_mean(power) / arithmetic_mean(power)
        # Ranges 0 (perfectly tonal) → 1 (white noise / flat spectrum)
        power = np.abs(np.fft.rfft(block)[1:]) ** 2   # drop DC bin
        power = np.maximum(power, 1e-10)               # guard against log(0)
        flatness = np.exp(np.mean(np.log(power))) / np.mean(power)
        flatness_values.append(flatness)

flatness_values = np.array(flatness_values)
print(f"Total windows: {len(flatness_values):,}")
print(f"\nSpectral flatness distribution (0=tonal, 1=flat/noise):")
for p in [5, 25, 50, 75, 95]:
    print(f"  {p}th %:  {np.percentile(flatness_values, p):.4f}")

# --- Find sustained high-flatness stretches (2+ minutes) ---
# Top 30% flatness = most background-like windows
flat_threshold = np.percentile(flatness_values, 70)
is_flat = flatness_values >= flat_threshold

min_run_windows = int(120 / window_sec)   # 2-minute minimum
flat_runs = []
i = 0
while i < len(is_flat):
    if is_flat[i]:
        j = i
        while j < len(is_flat) and is_flat[j]:
            j += 1
        if (j - i) >= min_run_windows:
            flat_runs.append((i, j, flatness_values[i:j].mean(), rms_values[i:j].mean()))
        i = j
    else:
        i += 1

print(f"\nFlatness threshold (top 30%): {flat_threshold:.4f}")
print(f"Found {len(flat_runs)} sustained flat stretches of >= 2 minutes")

# --- Report top 10, sorted by highest flatness ---
flat_runs_sorted = sorted(flat_runs, key=lambda r: r[2], reverse=True)

print(f"\nTop 10 flattest sustained stretches:")
print(f"{'#':>3}  {'Start time (clock)':>22}  {'Offset':>10}  {'Duration':>10}  {'Flatness':>10}  {'RMS':>10}")
for n, (start_idx, end_idx, mean_flat, mean_rms) in enumerate(flat_runs_sorted[:10], 1):
    offset_sec = start_idx * window_sec
    clock = rec_start + timedelta(seconds=offset_sec)
    dur_sec = (end_idx - start_idx) * window_sec
    h, rem = divmod(int(offset_sec), 3600)
    m, s = divmod(rem, 60)
    print(f"{n:>3}  {clock.strftime('%Y-%m-%d %H:%M:%S'):>22}  {h:>2}:{m:02d}:{s:02d}  {dur_sec/60:>7.1f} min  {mean_flat:>10.4f}  {mean_rms:>10.6f}")

# Use spot_check(offset_sec) from the cell above to listen to any candidate
```

```python
from datetime import timedelta

# Confirmed background windows from spot-check (candidate #1 rejected — clear bird sound)
# Each entry: (offset_sec_start, duration_sec, label)
confirmed_windows = [
    (40140, 126, "candidate_2"),   # 05:09:02, 2.1 min — faint/distant, BirdNET-invisible
    (39054, 156, "candidate_3"),   # 04:50:56, 2.6 min — faint/distant, BirdNET-invisible
]

# Parse clip timestamps from clip names for Audio_Moth_3
am3_mask = labels["Recorder"] == "Audio_Moth_3"
labels.loc[am3_mask, "clip_dt"] = pd.to_datetime(
    labels.loc[am3_mask, "clip_name"].str.extract(r"Audio_Moth_3_(\d{8}_\d{6})")[0],
    format="%Y%m%d_%H%M%S"
)

# Map each confirmed window to clips and label them
total_labeled = 0
for offset_sec, duration_sec, name in confirmed_windows:
    win_start = rec_start + timedelta(seconds=offset_sec)
    win_end   = rec_start + timedelta(seconds=offset_sec + duration_sec)

    mask = (
        am3_mask &
        (labels["clip_dt"] >= win_start) &
        (labels["clip_dt"] <  win_end) &
        (labels["meaningful"] == "unknown")   # never overwrite a confident label
    )
    n = mask.sum()
    labels.loc[mask, "meaningful"] = "not_meaningful"
    labels.loc[mask, "meaningful_source"] = "background_energy"
    print(f"{name}: {win_start} → {win_end}  |  {n} clips labeled")
    total_labeled += n

print(f"\nTotal newly labeled: {total_labeled}")
print()
print(labels["meaningful"].value_counts())
print()
print(labels["meaningful_source"].value_counts())

# Drop the temporary helper column and save
labels = labels.drop(columns=["clip_dt"], errors="ignore")
labels.to_csv(config.LABELS_PROGRESS_PATH, index=False)
print("\nSaved to:", config.LABELS_PROGRESS_PATH)
```

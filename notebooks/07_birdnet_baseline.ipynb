# BirdNET v2.4 Baseline

Load the BirdNET v2.4 SavedModel (Protobuf / FP32) and verify it runs correctly on a test clip.
This is the starting point for the PTQ int8 compression pipeline.

```python
from birdnet.model_loader import load

model = load("acoustic", "2.4", "pb")
print("Model loaded:", type(model).__name__)
```

```python
TEST_CLIP = "/Users/qian/KWF/Segmented_Foldered/Audio Moth 1/0_4999/Audio_Moth_1_20250317_093627.wav"

result = model.predict(TEST_CLIP, top_k=5)
df = result.to_dataframe()
print(df[["start_time", "end_time", "species_name", "confidence"]].to_string(index=False))
```
